# NBA — Modélisation ML

Trois modèles entraînés et comparés :
1. **XGBoost baseline** — modèle classique de référence
2. **Réseau de neurones classique** — intermédiaire
3. **Hybride XGBoost → QNN** — modèle quantique

**Cibles :**
- Classification : `home_win` → probabilité P(victoire domicile)
- Régression : `point_diff` → différentiel de points prédit

**Métriques :**
- Accuracy, AUC-ROC, Log Loss (classification)
- MAE, RMSE (régression)

````python
import pandas as pd
import numpy as np
import pickle
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.metrics import (
    accuracy_score, roc_auc_score, log_loss,
    mean_absolute_error, mean_squared_error,
    confusion_matrix, classification_report
)
from xgboost import XGBClassifier, XGBRegressor

print('Imports OK')
````

````python
# Chargement des données
X_train_xgb = pd.read_csv('data/nba_X_train_xgb.csv')
X_val_xgb   = pd.read_csv('data/nba_X_val_xgb.csv')
X_train_qnn = pd.read_csv('data/nba_X_train_qnn.csv').values
X_val_qnn   = pd.read_csv('data/nba_X_val_qnn.csv').values

y_train      = pd.read_csv('data/nba_y_train.csv').values.ravel()
y_val        = pd.read_csv('data/nba_y_val.csv').values.ravel()
y_train_diff = pd.read_csv('data/nba_y_train_diff.csv').values.ravel()
y_val_diff   = pd.read_csv('data/nba_y_val_diff.csv').values.ravel()

with open('data/nba_features_qnn.pkl', 'rb') as f:
    features_qnn = pickle.load(f)

print(f'X_train_xgb : {X_train_xgb.shape}')
print(f'X_train_qnn : {X_train_qnn.shape}')
print(f'y_train     : {y_train.shape}')
print(f'Features QNN: {features_qnn}')

# Dictionnaire pour stocker les résultats
results = {}
````

````python
# Fonction d'évaluation commune
def evaluate_classifier(name, y_true, y_pred_proba):
    """
    Évalue un classifieur et retourne les métriques principales.
    y_pred_proba : probabilités P(home_win)
    """
    y_pred = (y_pred_proba >= 0.5).astype(int)
    
    acc     = accuracy_score(y_true, y_pred)
    auc     = roc_auc_score(y_true, y_pred_proba)
    logloss = log_loss(y_true, y_pred_proba)
    
    results[name] = {'accuracy': acc, 'auc': auc, 'log_loss': logloss}
    
    print(f'\n=== {name} ===')
    print(f'  Accuracy : {acc:.3f} ({acc:.1%})')
    print(f'  AUC-ROC  : {auc:.3f}')
    print(f'  Log Loss : {logloss:.4f}')
    print(f'  Baseline (toujours domicile) : {y_true.mean():.1%}')
    
    return y_pred_proba

def evaluate_regressor(name, y_true, y_pred):
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    print(f'\n=== {name} (régression) ===')
    print(f'  MAE  : {mae:.2f} points')
    print(f'  RMSE : {rmse:.2f} points')
    return mae, rmse
````

## 1. XGBoost Baseline

Modèle de référence classique. C'est la performance qu'on cherche à dépasser avec le QNN.

````python
# Classification — victoire/défaite
xgb_clf = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='logloss',
    early_stopping_rounds=20,
    verbosity=0
)

xgb_clf.fit(
    X_train_xgb, y_train,
    eval_set=[(X_val_xgb, y_val)],
    verbose=False
)

xgb_proba = xgb_clf.predict_proba(X_val_xgb)[:, 1]
evaluate_classifier('XGBoost', y_val, xgb_proba)
````

````python
# Régression — point diff
xgb_reg = XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    random_state=42,
    verbosity=0
)
xgb_reg.fit(X_train_xgb, y_train_diff)
xgb_diff_pred = xgb_reg.predict(X_val_xgb)
evaluate_regressor('XGBoost', y_val_diff, xgb_diff_pred)

# Feature importance
fi = pd.Series(xgb_clf.feature_importances_,
               index=X_train_xgb.columns).sort_values(ascending=False)
plt.figure(figsize=(8, 5))
fi.head(15).plot(kind='barh', color='#f39c12')
plt.title('XGBoost — Feature Importance')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig('data/nba_xgb_importance.png', dpi=100, bbox_inches='tight')
plt.show()
````

````python
# Sauvegarder XGBoost
with open('data/nba_xgb_clf.pkl', 'wb') as f:
    pickle.dump(xgb_clf, f)
with open('data/nba_xgb_reg.pkl', 'wb') as f:
    pickle.dump(xgb_reg, f)
print('XGBoost sauvegardé')
````

## 2. Réseau de neurones classique (intermédiaire)

Avant le QNN, on teste un NN classique pour avoir une baseline neurale.

````python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

# Dataset PyTorch
X_tr = torch.tensor(X_train_qnn, dtype=torch.float32)
y_tr = torch.tensor(y_train, dtype=torch.float32)
X_v  = torch.tensor(X_val_qnn,   dtype=torch.float32)
y_v  = torch.tensor(y_val,       dtype=torch.float32)

train_ds = TensorDataset(X_tr, y_tr)
train_dl = DataLoader(train_ds, batch_size=64, shuffle=True)

# Architecture
class ClassicNN(nn.Module):
    def __init__(self, n_features):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_features, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x).squeeze()

nn_model = ClassicNN(X_train_qnn.shape[1])
optimizer = torch.optim.Adam(nn_model.parameters(), lr=1e-3, weight_decay=1e-4)
criterion = nn.BCELoss()

# Entraînement
EPOCHS = 100
train_losses, val_losses = [], []

for epoch in range(EPOCHS):
    nn_model.train()
    epoch_loss = 0
    for xb, yb in train_dl:
        optimizer.zero_grad()
        pred = nn_model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    
    nn_model.eval()
    with torch.no_grad():
        val_pred = nn_model(X_v)
        val_loss = criterion(val_pred, y_v).item()
    
    train_losses.append(epoch_loss / len(train_dl))
    val_losses.append(val_loss)
    
    if (epoch+1) % 20 == 0:
        print(f'Epoch {epoch+1:3d} — train_loss: {train_losses[-1]:.4f} val_loss: {val_losses[-1]:.4f}')

# Courbe d'apprentissage
plt.figure(figsize=(8, 4))
plt.plot(train_losses, label='Train')
plt.plot(val_losses,   label='Val')
plt.xlabel('Epoch')
plt.ylabel('BCE Loss')
plt.title('Courbe apprentissage NN classique')
plt.legend()
plt.tight_layout()
plt.savefig('data/nba_nn_learning_curve.png', dpi=100, bbox_inches='tight')
plt.show()
````

````python
nn_model.eval()
with torch.no_grad():
    nn_proba = nn_model(X_v).numpy()

evaluate_classifier('NN Classique', y_val, nn_proba)

with open('data/nba_nn_classic.pkl', 'wb') as f:
    pickle.dump(nn_model, f)
print('NN classique sauvegardé')
````

## 3. Modèle Hybride XGBoost → QNN

**Architecture :**
```
Features (16) → AngleEmbedding → VQC (BasicEntanglerLayers) → PauliZ → Sigmoid → P(win)
```

**Stratégie :**
- XGBoost réduit les features et détecte les interactions non-linéaires
- QNN exploite l'espace de Hilbert pour la classification finale

⚠️ Le QNN est lent à entraîner — commencez avec peu d'epochs pour tester.

````python
import pennylane as qml
from pennylane import numpy as pnp

N_QUBITS = X_train_qnn.shape[1]  # 16
N_LAYERS = 3                      # profondeur du circuit

print(f'Qubits : {N_QUBITS}')
print(f'Layers : {N_LAYERS}')
print(f'Paramètres : {N_LAYERS * N_QUBITS} angles')

# Device quantique (simulateur)
dev = qml.device('default.qubit', wires=N_QUBITS)

@qml.qnode(dev, interface='torch')
def quantum_circuit(features, weights):
    """
    Circuit quantique variationnel.
    - AngleEmbedding : encode les features comme angles de rotation Ry
    - BasicEntanglerLayers : couches variationnelles avec entanglement
    - PauliZ sur qubit 0 : mesure → valeur dans [-1, 1]
    """
    # Encoding des features
    qml.AngleEmbedding(features, wires=range(N_QUBITS), rotation='Y')
    
    # Couches variationnelles
    qml.BasicEntanglerLayers(weights, wires=range(N_QUBITS))
    
    # Mesure
    return qml.expval(qml.PauliZ(0))

# Visualiser le circuit sur un exemple
sample_weights = pnp.random.uniform(0, np.pi, (N_LAYERS, N_QUBITS))
sample_features = X_train_qnn[0]
print('\nCircuit quantique :')
print(qml.draw(quantum_circuit)(sample_features, sample_weights))
````

````python
# Modèle hybride PyTorch + PennyLane
import torch
import torch.nn as nn

class HybridQNN(nn.Module):
    """
    Modèle hybride classique-quantique.
    
    Couche classique optionnelle (compression) + circuit quantique.
    La sortie PauliZ ∈ [-1,1] est passée dans sigmoid → P(home_win).
    """
    def __init__(self, n_qubits, n_layers):
        super().__init__()
        self.n_qubits = n_qubits
        
        # Poids variationnels du circuit quantique
        self.weights = nn.Parameter(
            torch.tensor(
                np.random.uniform(0, np.pi, (n_layers, n_qubits)),
                dtype=torch.float32
            )
        )
        
        # Couche de scaling optionnelle
        self.scaling = nn.Parameter(torch.ones(1))
        self.bias    = nn.Parameter(torch.zeros(1))

    def forward(self, x):
        # Passer chaque sample dans le circuit quantique
        out = torch.stack([
            quantum_circuit(x[i], self.weights)
            for i in range(x.shape[0])
        ])
        # Transformer [-1,1] → probabilité [0,1]
        out = self.scaling * out + self.bias
        return torch.sigmoid(out)

qnn_model = HybridQNN(N_QUBITS, N_LAYERS)
print(f'Paramètres entraînables : {sum(p.numel() for p in qnn_model.parameters())}')
````

````python
# Entraînement QNN
# ⚠️ Lent : ~10-30 sec par epoch sur simulateur
# Commencer avec EPOCHS=10 pour tester, puis augmenter

EPOCHS_QNN  = 30        # ← augmenter si résultats corrects
BATCH_SIZE  = 32        # ← petit batch pour le QNN
LR_QNN      = 0.01

X_tr_q = torch.tensor(X_train_qnn, dtype=torch.float32)
y_tr_q = torch.tensor(y_train, dtype=torch.float32)
X_v_q  = torch.tensor(X_val_qnn,   dtype=torch.float32)
y_v_q  = torch.tensor(y_val,       dtype=torch.float32)

train_ds_q = TensorDataset(X_tr_q, y_tr_q)
train_dl_q = DataLoader(train_ds_q, batch_size=BATCH_SIZE, shuffle=True)

optimizer_q = torch.optim.Adam(qnn_model.parameters(), lr=LR_QNN)
criterion_q = nn.BCELoss()

qnn_train_losses, qnn_val_losses = [], []

for epoch in range(EPOCHS_QNN):
    qnn_model.train()
    epoch_loss = 0
    for xb, yb in train_dl_q:
        optimizer_q.zero_grad()
        pred = qnn_model(xb)
        loss = criterion_q(pred, yb)
        loss.backward()
        optimizer_q.step()
        epoch_loss += loss.item()

    qnn_model.eval()
    with torch.no_grad():
        val_pred_q = qnn_model(X_v_q)
        val_loss_q = criterion_q(val_pred_q, y_v_q).item()

    qnn_train_losses.append(epoch_loss / len(train_dl_q))
    qnn_val_losses.append(val_loss_q)

    print(f'Epoch {epoch+1:3d}/{EPOCHS_QNN} — '
          f'train: {qnn_train_losses[-1]:.4f} '
          f'val: {qnn_val_losses[-1]:.4f}')
````

````python
qnn_model.eval()
with torch.no_grad():
    qnn_proba = qnn_model(X_v_q).numpy()

evaluate_classifier('QNN Hybride', y_val, qnn_proba)

torch.save(qnn_model.state_dict(), 'data/nba_qnn_weights.pt')
print('QNN sauvegardé')
````

## 5. Fonction de prédiction — Interface production

Fonction utilisée par le LLM pour prédire un match en production.

````python
def predict_game(home_team, away_team,
                 home_rest_days=2, away_rest_days=2,
                 home_is_b2b=0, away_is_b2b=0,
                 absent_players_home=None, absent_players_away=None,
                 model='xgboost'):
    """
    Prédit le résultat d'un match NBA.
    
    Paramètres
    ----------
    home_team : str  ex: 'LAL'
    away_team : str  ex: 'GSW'
    home_rest_days : int  jours de repos équipe domicile
    away_rest_days : int  jours de repos équipe extérieure
    home_is_b2b : int  1 si back-to-back domicile
    away_is_b2b : int  1 si back-to-back extérieur
    absent_players_home : list  ex: ['LeBron James', 'Anthony Davis']
    absent_players_away : list  ex: ['Stephen Curry']
    model : str  'xgboost' | 'qnn'
    
    Retourne
    --------
    dict avec probabilités et prédiction
    """
    import pickle
    
    # Charger les stats actuelles des équipes
    team_stats = pd.read_csv('data/nba_team_stats_current.csv')
    team_recent = pd.read_csv('data/nba_team_stats_recent.csv')
    player_stats = pd.read_csv('data/nba_player_stats_current.csv')
    
    # Récupérer stats home et away
    home_stats = team_stats[team_stats['TEAM_ABBREVIATION'] == home_team]
    away_stats = team_stats[team_stats['TEAM_ABBREVIATION'] == away_team]
    
    if home_stats.empty or away_stats.empty:
        available = team_stats['TEAM_ABBREVIATION'].tolist()
        raise ValueError(f'Équipe non trouvée. Équipes disponibles : {available}')
    
    home_stats = home_stats.iloc[0]
    away_stats = away_stats.iloc[0]
    
    # Construire le vecteur de features
    with open('data/nba_features_qnn.pkl', 'rb') as f:
        feature_names = pickle.load(f)
    with open('data/nba_scaler_qnn.pkl', 'rb') as f:
        scaler = pickle.load(f)
    
    # Mapping des features depuis les stats équipes
    STAT_MAP = {
        'net_rating_diff':                 home_stats.get('NET_RATING', 0) - away_stats.get('NET_RATING', 0),
        'efg_diff':                        home_stats.get('EFG_PCT', 0) - away_stats.get('EFG_PCT', 0),
        'adv_home_estimatedDefensiveRating': home_stats.get('DEF_RATING', 100),
        'adv_home_estimatedOffensiveRating': home_stats.get('OFF_RATING', 100),
        'home_FG3_PCT':                    home_stats.get('FG3_PCT', 0.35),
        'away_FG3_PCT':                    away_stats.get('FG3_PCT', 0.35),
        'home_AST':                        home_stats.get('AST', 25),
        'away_AST':                        away_stats.get('AST', 25),
        'home_REB':                        home_stats.get('REB', 44),
        'away_REB':                        away_stats.get('REB', 44),
        'home_TOV':                        home_stats.get('TOV', 14),
        'home_FT_PCT':                     home_stats.get('FT_PCT', 0.77),
        'away_FT_PCT':                     away_stats.get('FT_PCT', 0.77),
        'pt_diff_diff_10':                 home_stats.get('PLUS_MINUS', 0) - away_stats.get('PLUS_MINUS', 0),
        'home_roll_pt_diff_10':            home_stats.get('PLUS_MINUS', 0),
        'away_roll_pt_diff_10':            away_stats.get('PLUS_MINUS', 0),
    }
    
    # Construire le vecteur dans le bon ordre
    x = np.array([STAT_MAP.get(f, 0) for f in feature_names]).reshape(1, -1)
    x_scaled = scaler.transform(x)
    x_scaled = np.clip(x_scaled, -np.pi, np.pi)
    
    # Prédiction
    if model == 'xgboost':
        with open('data/nba_scaler_xgb.pkl', 'rb') as f:
            scaler_xgb = pickle.load(f)
        with open('data/nba_features_xgb.pkl', 'rb') as f:
            features_xgb = pickle.load(f)
        with open('data/nba_xgb_clf.pkl', 'rb') as f:
            clf = pickle.load(f)
        x_xgb = np.array([STAT_MAP.get(f, 0) for f in features_xgb]).reshape(1, -1)
        x_xgb_scaled = scaler_xgb.transform(x_xgb)
        p_home = clf.predict_proba(x_xgb_scaled)[0, 1]
    else:
        x_t = torch.tensor(x_scaled, dtype=torch.float32)
        qnn_model.eval()
        with torch.no_grad():
            p_home = float(qnn_model(x_t)[0])
    
    p_away = 1 - p_home
    winner = home_team if p_home > 0.5 else away_team
    confidence = max(p_home, p_away)
    
    result = {
        'home_team':    home_team,
        'away_team':    away_team,
        'p_home_win':   round(p_home, 3),
        'p_away_win':   round(p_away, 3),
        'predicted_winner': winner,
        'confidence':   round(confidence, 3),
        'model':        model,
    }
    
    print(f'\n🏀 {home_team} vs {away_team}')
    print(f'   P({home_team} gagne) = {p_home:.1%}')
    print(f'   P({away_team} gagne) = {p_away:.1%}')
    print(f'   → Prédiction : {winner} ({confidence:.1%} de confiance)')
    
    return result

# Test
predict_game('LAL', 'GSW', model='xgboost')
````

````python
# Test avec QNN
predict_game('BOS', 'MIA', model='qnn')
````
