```python
def process_fbref_team_data(df):
    """
    Pipeline complète : Renommage, Extraction du club, Sélection et 
    Normalisation des statistiques (division par 38).
    """
    
    # 1. Renommage des colonnes (Fusion des deux niveaux d'en-tête)
    nouveaux_noms = []
    for nom_colonne, sous_categorie in zip(df.columns, df.iloc[0]):
        # On retire le suffixe .1, .2 ajouté par pandas
        nom_principal = nom_colonne.split('.')[0] 
        
        if pd.isna(sous_categorie):
            nouveaux_noms.append(nom_principal)
        else:
            nouveaux_noms.append(f"{nom_principal}_{sous_categorie}")
            
    df.columns = nouveaux_noms

    # 2. Suppression de la ligne 0 (sous-catégories) et reset de l'index
    df = df.iloc[1:].reset_index(drop=True)

    # 3. Extraction du nom du club depuis la colonne 'url'
    # Logique : on prend la partie après le dernier '/', puis on retire '-Stats'
    if 'url' in df.columns:
        df['club'] = df['url'].str.split('/').str[-1].str.replace('-Stats', '', regex=False)

    # 4. Sélection des colonnes cibles
    # Note : On inclut 'club' dans la sélection
    colonnes_a_garder = [
        "club", "players_used", "Age", "Poss", 
        "Per 90 Minutes_Gls", "Per 90 Minutes_Ast", 
        "Performance_PK", "Performance_PKatt", 
        "Performance_CrdY", "Performance_CrdR"
    ]
    
    # On filtre en vérifiant que les colonnes existent bien
    df = df[[c for c in colonnes_a_garder if c in df.columns]].copy()

    # 5. Conversion en numérique et division par 38
    # On cible les PK et les Cartons (CrdY et CrdR)
    cols_a_transformer = [
        'Performance_PK', 'Performance_PKatt', 
        'Performance_CrdY', 'Performance_CrdR'
    ]
    
    for col in cols_a_transformer:
        if col in df.columns:
            # On force la conversion en nombre avant de diviser
            df[col] = pd.to_numeric(df[col], errors='coerce') / 38
            
    return df

# --- Utilisation ---
fbref_team2024 = process_fbref_team_data(fbref_team2024)
fbref_team2025 = process_fbref_team_data(fbref_team2025)
fbref_team2024["season"] = "2024-2025"
fbref_team2025["season"] = "2025-2026"
```

```python
# ---------------------------------------------------------
# ÉTAPE 0 : Correction des anomalies dans la source pl_events
# ---------------------------------------------------------
pl_events['home_team'] = pl_events['home_team'].replace({'Bolton Wanderers': 'Wolverhampton'})
pl_events['away_team'] = pl_events['away_team'].replace({'Bolton Wanderers': 'Wolverhampton'})

# ---------------------------------------------------------
# ÉTAPE 1 : Uniformiser les colonnes de jointure
# ---------------------------------------------------------
fdcu = fdcu.rename(columns={'HomeTeam': 'home_team', 'AwayTeam': 'away_team', 'Date': 'date'})
pl_events = pl_events.rename(columns={'event_date' : 'date'})

# ---------------------------------------------------------
# ÉTAPE 2 : Créer les dictionnaires de correction
# ---------------------------------------------------------
map_fbref = {
    'Manchester Utd': 'Manchester United',
    'Brighton': 'Brighton & Hove Albion',
    'Wolves': 'Wolverhampton'
}

map_fdcu = {
    'Man United': 'Manchester United',
    'Ipswich': 'Ipswich Town',
    "Nott'm Forest": 'Nottingham Forest',
    'Newcastle': 'Newcastle United',
    'West Ham': 'West Ham United',
    'Leicester': 'Leicester City',
    'Brighton': 'Brighton & Hove Albion',
    'Man City': 'Manchester City',
    'Tottenham': 'Tottenham Hotspur',
    'Wolves': 'Wolverhampton',
    'Leeds': 'Leeds United'
}

# ---------------------------------------------------------
# ÉTAPE 3 : Fonction de nettoyage universelle
# ---------------------------------------------------------
def clean_and_map(df, mapping_dict=None):
    df_clean = df.copy()
    
    if mapping_dict:
        df_clean['home_team'] = df_clean['home_team'].replace(mapping_dict)
        df_clean['away_team'] = df_clean['away_team'].replace(mapping_dict)
    
    df_clean['home_team'] = df_clean['home_team'].str.strip()
    df_clean['away_team'] = df_clean['away_team'].str.strip()
    df_clean['date'] = pd.to_datetime(df_clean['date']).dt.normalize()
    
    return df_clean

# Application du nettoyage
pl_events_clean = clean_and_map(pl_events, mapping_dict=None)
fbref_clean     = clean_and_map(fbref, mapping_dict=map_fbref)
fdcu_clean      = clean_and_map(fdcu, mapping_dict=map_fdcu)

# ---------------------------------------------------------
# ÉTAPE 4 : Fusion avec tolérance (+/- 1 jour) et priorité FDCU
# ---------------------------------------------------------
# 1. Fusion pl_events + fbref 
merged_1 = pd.merge(
    pl_events_clean, 
    fbref_clean, 
    on=['home_team', 'away_team'], 
    how='inner',
    suffixes=('', '_fbref')
)
merged_1 = merged_1[(merged_1['date'] - merged_1['date_fbref']).abs() <= pd.Timedelta(days=1)]
merged_1 = merged_1.drop(columns=['date_fbref']) 

# 2. Fusion (pl_events + fbref) + fdcu
final_df = pd.merge(
    merged_1, 
    fdcu_clean, 
    on=['home_team', 'away_team'], 
    how='inner', 
    suffixes=('', '_fdcu')
)
final_df = final_df[(final_df['date'] - final_df['date_fdcu']).abs() <= pd.Timedelta(days=1)]

# 3. CONSERVATION DE LA DATE FDCU
final_df['date'] = final_df['date_fdcu']
final_df = final_df.drop(columns=['date_fdcu'])

# ---------------------------------------------------------
# VÉRIFICATION GLOBALE
# ---------------------------------------------------------
print(f"Lignes dans pl_events : {len(pl_events_clean)}")
print(f"Lignes dans fbref     : {len(fbref_clean)}")
print(f"Lignes dans fdcu      : {len(fdcu_clean)}")
print("-" * 30)
print(f"Lignes après fusion   : {len(final_df)}")
```

```python
# 1. Création d'un dictionnaire de correspondance spécifique pour FBRef
# On gère les cas particuliers (tirets, "&", suffixes "Wanderers", etc.)
mapping_fbref_clubs = {
    'Brighton-and-Hove-Albion': 'Brighton & Hove Albion',
    'Wolverhampton-Wanderers': 'Wolverhampton',
    'Aston-Villa': 'Aston Villa',
    'Crystal-Palace': 'Crystal Palace',
    'Ipswich-Town': 'Ipswich Town',
    'Leicester-City': 'Leicester City',
    'Manchester-City': 'Manchester City',
    'Manchester-United': 'Manchester United',
    'Newcastle-United': 'Newcastle United',
    'Nottingham-Forest': 'Nottingham Forest',
    'Tottenham-Hotspur': 'Tottenham Hotspur',
    'West-Ham-United': 'West Ham United',
    'Leeds-United': 'Leeds United',
    'Bournemouth': 'Bournemouth',
    'Southampton': 'Southampton'
}

# Fonction pour normaliser les noms fbref_team['club']
def normalize_fbref_team_name(name):
    # Si le nom est dans notre dico d'exceptions
    if name in mapping_fbref_clubs:
        return mapping_fbref_clubs[name]
    # Sinon, on remplace simplement les tirets par des espaces
    return name.replace('-', ' ')

# Appliquer la normalisation sur le dataset des équipes
fbref_team['club_clean'] = fbref_team['club'].apply(normalize_fbref_team_name)

# 2. Identification des colonnes de statistiques à ramener
# On exclut les clés de jointure ('club', 'club_clean', 'season_id')
stats_cols = [c for c in fbref_team.columns if c not in ['club', 'club_clean', 'season_id']]

# ---------------------------------------------------------
# 3. FUSION POUR L'ÉQUIPE DOMICILE (HOME)
# ---------------------------------------------------------
final_df = pd.merge(
    final_df,
    fbref_team,
    left_on=['home_team', 'season_id'],
    right_on=['club_clean', 'season_id'],
    how='left'
)

# Renommer les colonnes ajoutées avec le préfixe 'home_'
final_df = final_df.rename(columns={c: f'home_{c}' for c in stats_cols})
# Supprimer les colonnes de jointure inutiles après merge
final_df = final_df.drop(columns=['club', 'club_clean'])

# ---------------------------------------------------------
# 4. FUSION POUR L'ÉQUIPE EXTÉRIEURE (AWAY)
# ---------------------------------------------------------
final_df = pd.merge(
    final_df,
    fbref_team,
    left_on=['away_team', 'season_id'],
    right_on=['club_clean', 'season_id'],
    how='left'
)

# Renommer les colonnes ajoutées avec le préfixe 'away_'
final_df = final_df.rename(columns={c: f'away_{c}' for c in stats_cols})
# Supprimer les colonnes de jointure inutiles après merge
final_df = final_df.drop(columns=['club', 'club_clean'])

print(f"Fusion terminée !  {len(final_df.shape)}")
```

```python
# ---------------------------------------------------------
# ÉTAPE 1 : Nettoyage et harmonisation du DataFrame Elo
# ---------------------------------------------------------

# Dictionnaire pour convertir les noms Elo vers les standards de final_df
map_elo = {
    'Man City': 'Manchester City',
    'Man United': 'Manchester United',
    'Brighton': 'Brighton & Hove Albion',
    'Newcastle': 'Newcastle United',
    'Forest': 'Nottingham Forest',
    'Tottenham': 'Tottenham Hotspur',
    'Wolves': 'Wolverhampton',
    'Ipswich': 'Ipswich Town',
    'Leeds': 'Leeds United',
    'West Ham': 'West Ham United',
    'Leicester': 'Leicester City',
    'Southampton': 'Southampton'
    # Ajoute d'autres correspondances si tu croises d'autres anomalies
}

# Appliquer le nom propre
clubelo['Club_clean'] = clubelo['Club'].replace(map_elo)

# S'assurer que les dates sont bien au format datetime pour la comparaison
clubelo['From'] = pd.to_datetime(clubelo['From']).dt.normalize()
clubelo['To'] = pd.to_datetime(clubelo['To']).dt.normalize()
final_df['date'] = pd.to_datetime(final_df['date']).dt.normalize()


# ---------------------------------------------------------
# ÉTAPE 2 : Fonction de recherche (Lookup) par inégalité
# ---------------------------------------------------------
def get_elo_for_match(team_name, match_date):
    # 1. On filtre le dataframe Elo pour ne garder que l'équipe demandée
    team_elo = clubelo[clubelo['Club_clean'] == team_name]
    
    # 2. On cherche la ligne où la date du match est comprise entre From et To
    mask = (team_elo['From'] <= match_date) & (team_elo['To'] >= match_date)
    result = team_elo[mask]
    
    # 3. Si on trouve une correspondance, on renvoie le score Elo, sinon NaN
    if not result.empty:
        return result['Elo'].iloc[0]
    else:
        return None # Aucun Elo trouvé pour cette date


# ---------------------------------------------------------
# ÉTAPE 3 : Application sur final_df (Home et Away)
# ---------------------------------------------------------

# Appliquer la fonction pour l'équipe à domicile (axis=1 permet d'envoyer la ligne entière)
final_df['home_elo'] = final_df.apply(
    lambda row: get_elo_for_match(row['home_team'], row['date']), 
    axis=1
)

# Appliquer la fonction pour l'équipe à l'extérieur
final_df['away_elo'] = final_df.apply(
    lambda row: get_elo_for_match(row['away_team'], row['date']), 
    axis=1
)

# Vérification rapide des valeurs manquantes
print("Matchs sans Elo Domicile :", final_df['home_elo'].isna().sum())
print("Matchs sans Elo Extérieur :", final_df['away_elo'].isna().sum())
```

```python
# Créer un masque pour les valeurs manquantes (Domicile OU Extérieur)
mask_missing_elo = final_df['home_elo'].isna() | final_df['away_elo'].isna()

# Filtrer le dataframe
matchs_sans_elo = final_df[mask_missing_elo]

# Sélectionner uniquement les colonnes pertinentes pour y voir clair
colonnes_diagnostic = ['date', 'home_team', 'away_team', 'home_elo', 'away_elo']

print(f"Total de matchs avec un problème d'Elo : {len(matchs_sans_elo)}")
print("-" * 50)
matchs_sans_elo[colonnes_diagnostic]
```

```python
# ==========================================
# ÉTAPE 9.1 : Chargement & Fusion Historiques FDCU
# ==========================================


df_history = pd.concat([
        pd.read_csv("data/pl_fdcu_20_21.csv"),
        pd.read_csv("data/pl_fdcu_21_22.csv"),
        pd.read_csv("data/pl_fdcu_22_23.csv"),
        pd.read_csv("data/pl_fdcu_23_24.csv"),
        pd.read_csv("data/pl_fdcu24_25.csv"),
        pd.read_csv("data/pl_fdcu25_26.csv")
], ignore_index=True)

# Nettoyage date
df_history['Date'] = pd.to_datetime(df_history['Date'], format='%d/%m/%Y', errors='coerce').dt.normalize()
df_history = df_history.sort_values('Date').reset_index(drop=True)

# Mapping équipes
map_fdcu_dyn = {
    'Man United': 'Manchester United',
    'Ipswich': 'Ipswich Town',
    "Nott'm Forest": 'Nottingham Forest',
    'Newcastle': 'Newcastle United',
    'West Ham': 'West Ham United',
    'Leicester': 'Leicester City',
    'Brighton': 'Brighton & Hove Albion',
    'Man City': 'Manchester City',
    'Tottenham': 'Tottenham Hotspur',
    'Wolves': 'Wolverhampton',
    'Leeds': 'Leeds United'
}
df_history['HomeTeam'] = df_history['HomeTeam'].replace(map_fdcu_dyn)
df_history['AwayTeam'] = df_history['AwayTeam'].replace(map_fdcu_dyn)

print("✅ Historiques FDCU chargés")
```

```python
# ==========================================
# ÉTAPE 9.2 : Calcul des Dynamiques Temporelles
# ==========================================

df_history['home_rest_days'] = 7.0
df_history['away_rest_days'] = 7.0
df_history['home_points_last_5'] = 0.0
df_history['away_points_last_5'] = 0.0
df_history['h2h_home_win_pct'] = 0.5

team_last_match_date = {}
team_match_history = {team: [] for team in pd.concat([df_history['HomeTeam'], df_history['AwayTeam']]).unique()}
h2h_history = {}

print("⏳ Calcul des dynamiques temporelles...")

for index, row in df_history.iterrows():
    home = row['HomeTeam']
    away = row['AwayTeam']
    match_date = row['Date']
    
    # Fatigue
    if home in team_last_match_date:
        df_history.at[index, 'home_rest_days'] = (match_date - team_last_match_date[home]).days
    if away in team_last_match_date:
        df_history.at[index, 'away_rest_days'] = (match_date - team_last_match_date[away]).days
    team_last_match_date[home] = match_date
    team_last_match_date[away] = match_date

    # Forme
    home_past = team_match_history[home][-5:]
    away_past = team_match_history[away][-5:]
    df_history.at[index, 'home_points_last_5'] = sum(home_past) if len(home_past) > 0 else 0
    df_history.at[index, 'away_points_last_5'] = sum(away_past) if len(away_past) > 0 else 0
    
    result = row['FTR']
    if result == 'H':
        team_match_history[home].append(3)
        team_match_history[away].append(0)
    elif result == 'A':
        team_match_history[home].append(0)
        team_match_history[away].append(3)
    else:
        team_match_history[home].append(1)
        team_match_history[away].append(1)

    # Head-to-Head
    pair = tuple(sorted([home, away]))
    if pair not in h2h_history:
        h2h_history[pair] = []
    past_encounters = h2h_history[pair][-5:]
    if len(past_encounters) > 0:
        wins = sum(1 for encounter in past_encounters if encounter['winner'] == home)
        df_history.at[index, 'h2h_home_win_pct'] = wins / len(past_encounters)
    winner_name = home if result == 'H' else (away if result == 'A' else 'Draw')
    h2h_history[pair].append({'date': match_date, 'winner': winner_name})

print("✅ Dynamiques temporelles calculées")
```

```python
# ==========================================
# ÉTAPE 9.3 : Super-Features + Fusion Finale
# ==========================================

# Derby
liste_derbies = [
    {"Aston Villa", "Birmingham City"}, {"West Ham United", "Millwall"},
    {"Arsenal", "Tottenham Hotspur"}, {"Liverpool", "Everton"},
    {"Manchester United", "Manchester City"}, {"Sheffield United", "Sheffield Wednesday"},
    {"Bristol City", "Bristol Rovers"}, {"Stoke City", "Port Vale"},
    {"Nottingham Forest", "Notts County"}, {"Arsenal", "Chelsea"},
    {"Chelsea", "Fulham"}, {"Newcastle United", "Sunderland"},
    {"Ipswich Town", "Norwich City"}, {"Portsmouth", "Southampton"},
    {"Blackburn Rovers", "Burnley"}, {"West Bromwich Albion", "Wolverhampton"},
    {"Arsenal", "Manchester United"}, {"Chelsea", "Leeds United"},
    {"Leeds United", "Manchester United"}, {"Liverpool", "Manchester United"},
    {"Liverpool", "Manchester City"}
]

def est_un_derby(home, away):
    return 1 if {home, away} in liste_derbies else 0

pl_events['is_derby'] = pl_events.apply(lambda row: est_un_derby(row['home_team'], row['away_team']), axis=1)

# Blessures
def calculer_diff_absences(json_string):
    if pd.isna(json_string):
        return 0
    try:
        clean_json = str(json_string).replace('""', '"')
        if clean_json.startswith('"') and clean_json.endswith('"'):
            clean_json = clean_json[1:-1]
        data = json.loads(clean_json)
        return len(data.get('home', [])) - len(data.get('away', []))
    except:
        return 0

print("🏥 Extraction des blessures...")
pl_events['diff_absences'] = pl_events['unavailable_players'].apply(calculer_diff_absences)

# Momentum
historique_momentum = {team: [] for team in pd.concat([pl_events['home_team'], pl_events['away_team']]).unique()}
pl_events['home_momentum'] = 0.0
pl_events['away_momentum'] = 0.0

print("📈 Calcul du Momentum...")
for index, row in pl_events.iterrows():
    home = row['home_team']
    away = row['away_team']
    past_home = historique_momentum[home][-5:]
    past_away = historique_momentum[away][-5:]
    pl_events.at[index, 'home_momentum'] = sum(past_home) if len(past_home) > 0 else 0
    pl_events.at[index, 'away_momentum'] = sum(past_away) if len(past_away) > 0 else 0
    if not pd.isna(row['home_score']) and not pd.isna(row['actual_home_xg']):
        historique_momentum[home].append(row['home_score'] - row['actual_home_xg'])
    if not pd.isna(row['away_score']) and not pd.isna(row['actual_away_xg']):
        historique_momentum[away].append(row['away_score'] - row['actual_away_xg'])

print("✅ Super-features calculées")
```
